### Weather Temperature Analysis with NumPy

```python
# Creating a python list and Numpy array 
temperatures = [27.5, 29.0, 30.5, 28.00, 31.0, 32.5, 30.0]
temperatures_array = np.array(temperatures)
print(temperatures_array)
# Ans 1. The temperature is a 1D array
# Ans 2. It is a Vector
# Ans 3. The data type is and integer
# Counting the number in temperatures observed
temperatures_obeserved = len(temperatures)
print("Nunbers Of Temperatures Obeserved: ", temperatures_obeserved)
```

```python
# Q1: Is it 1D or 2D?
print("Dimensions:", temperatures_array.ndim) # 1D

# Q2: Vector or Matrix?
# Answer: Vector (1D array is called a vector)

# Q3: What is its data type?
print("dtype:", temperatures_array.dtype) # float64

# Q4: How many observations?
temperatures_obeserved = len(temperatures) #7
print("Nunbers Of Temperatures Obeserved: ", temperatures_obeserved)
```

```python
# calculating the difference between the rocorded temperature and the targeted temperature
target = [28.0, 28.0, 29.0, 28.0, 30.0, 30.0, 29.0]
target_array = np.array(target)
temperature_difference = temperatures_array - target_array
# Increasing the temperature by 2 degree census
temperature_increase = (2 + temperature_difference)
# No for loop needed! NumPy broadcasts the +2 to every element.

# displaying output
print("Temperature Difference:", temperature_difference)
print()
print("Temperature Increase:", temperature_increase)
```

# MINI PROJECT CHALLENGE

```python
# creating Python list and Numpy array for three location
lagos = [27.5, 29.0, 30.5, 28.0, 30.0]
abuja = [25.0, 28.0, 29.0, 27.5, 30.5]
ibadan = [26.5, 28.5, 29.0, 27.5, 30.5]

lagos_array = np.array(lagos)
abuja_array = np.array(abuja)
ibadan_array = np.array(ibadan)

# Stack into a 2D array for easier multi-location analysis (shape: 3 locations x 5 days)
all_locations = np.array([lagos, abuja, ibadan])
locations = ["Lagos", "Abuja", "Ibadan"]
# Displaying output
print("Lagos Temperature:", lagos_array)
print("Abuja Temperature:", abuja_array)
print("Ibadan Temperature:", ibadan_array)
```

```python
# Which location recorded the highest temperature on each day
max_indices_per_day = np.argmax(all_locations, axis=0)
highest_location_per_day = [locations[idx] for idx in max_indices_per_day]

print("Daily max location:", highest_location_per_day)
```

```python
# Increasing every recorded temperature by 1 degree census
all_plus_1 = all_locations + 1
print("All locations (+1°C):\n", all_plus_1)
```

```python
# Comparing Lagos temperatures with Abuja temperatures
temperature_difference = lagos_array - abuja_array
print("Difference (Lagos - Abuja):", temperature_difference)
```
